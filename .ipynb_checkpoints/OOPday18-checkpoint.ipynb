```python
# assignment

# Fill in the Line class so it accepts two coordinate
# tuples and can return the line's slope and distance.
class Line:
    def __init__(self,coor1,coor2):
        self.x = coor1
        self.y = coor2
    def distance(self):
        dis =(((self.y[0]-self.x[0])**2) +((self.y[1]-self.x[1])**2))**0.5
        return dis
    def slope(self):
        sl = ((self.y[1]-self.x[1])/(self.y[0]-self.x[0]))
        return sl
```

```python
l1 = Line(coor1=(3,2), coor2=(8,10))
```

```python
print("The distance is: ",l1.distance())
```

```python
print("The slope is: ",l1.slope())
```

```python
# Fill in the Cylinder class.
class Cylinder:
    pi = 3.14
    def __init__(self, height=1,radius=1):
        self.h = height
        self.r = radius
    def volume(self):
        vol_cyl = (self.pi)*((self.r)**2)*(self.h)
        return vol_cyl
    def surface_area(self):
        sur_area = 2*(self.pi)*(self.r)*(self.r + self.h)
        return sur_area
```

```python
# example output
cyl = Cylinder(2,3)
```

```python
cyl.volume()
```

```python
cyl.surface_area()
```

```python
# Fill in the Account class. It should keep track of an owner and a balance,
# support deposits and withdrawals (a withdrawal larger than the balance 
# should print 'Insufficient funds' and not change the balance), and printing
# theaccount should display the owner and balance in a readable way.

class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
    def __str__(self):
        return f"Account owner is {self.owner} and Account balance is {self.balance}."
    def deposit(self, dep_amt):
        self.balance += dep_amt
        print(f"Amount {dep_amt} deposited.")
    def withdraw(self, wd_amt):
        if wd_amt > self.balance:
            print(f"Balance is {self.balance} . So, Insufficient funds")
        else:
            self.balance -= wd_amt
            print(f"Amount {wd_amt} Withdrawal from your account.")
```

```python
# EXAMPLE OUTPUT
acct1 = Account('Jose', 100)
```

```python
print(acct1)
```

```python
acct1.deposit(50)
```

```python
acct1.withdraw(75)
```

```python
acct1.withdraw(500)
```

```python
# Fill in the Rectangle class so it accepts a length and width and
# can return the rectangle's area and perimeter.

class Rectangle:
    def __init__(self, length=1, width=1):
        self.l = length
        self.w = width
    def area(self):
        area_rec = self.l * self.w
        return area_rec
    def perimeter(self):
        perm_rec = 2*(self.l + self.w)
        return perm_rec
```

```python
# EXAMPLE OUTPUT
rect = Rectangle(4, 7)
```

```python
rect.area()
```

```python
rect.perimeter()
```

```python
# Fill in the Animal class and have the Dog class inherit from it. Animal should store 
# a name and have a generic speak() method. Dog should override speak() so it 
# returns a bark specific to the dog's name.

class Animal:
    def __init__(self, name):
        self.nam_anim = name
    def speak(self):
        pass
        
class Dog(Animal):
    def speak(self):
        return f"{self.nam_anim} barks."
```

```python
# EXAMPLE OUTPUT
d = Dog('Rex')
```

```python
d.speak()
```
