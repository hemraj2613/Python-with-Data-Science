```python
# polymorphism--> same method multiple forms
class Animal:
    def sound(self):
        return "Animal generate sound."
class Dog(Animal):
    def sound(self):
        return "Dog barks."
class Cat(Animal):
    def sound(self):
        return "Cat meows."

animals = [Dog(),Cat(),Animal()]
for animal in animals:
    print(animal.sound())
```

```python
# Abstraction ==> To hide inner logic
# don't implement anything
# only define methods
# it only can be inherited
# objects of abstact class cannot be made
```

```python
from abc import ABC, abstractmethod
```

```python
class Bank(ABC):
    @abstractmethod
    def secure(self):
        pass
    @abstractmethod
    def privacy(self):
        pass
    @abstractmethod
    def interest_rate(self):
        pass
        
class Esewa(Bank):
    def secure(self):
        return "Security provided."
    def privacy(self):
        return "Privacy provided."
    def interest_rate(self):
        return "Interest_rate provided."
        
esewa = Esewa()
esewa.secure()
```

```python
esewa.privacy()
```

```python
esewa.interest_rate()
```

```python
# encapsulation
class Student:
    def __init__(self,name,password):
        self.name = name  #public
        self.__password = password  #private(only accessible in another method)
    def details(self):
        return f"Student name is {self.name} and password is {self.__password}."
student = Student("hem","Hem@123")
print(student.name)
```

```python
print(student.details())
```
