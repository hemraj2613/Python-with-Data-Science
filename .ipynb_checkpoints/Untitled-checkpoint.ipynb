```python
# shallow copy and deep copy
a = 10
id(a)
```

```python
b = 10
id(b)
```

```python
# shallow copy
import copy
a= [1,2,3,5]
a[0]=9
b = a.copy()
print(a)
print(b)
print(id(a))
print(id(b))
```

```python
# deep copy
import copy
a= [1,2,3,5]
b = copy.deepcopy(a)
a[0]=9
print(a)
print(b)
```
