```python
# store a valid username and password in a dictionary
user_credentials = {
    "username" : "Hemraj123",
    "password" : "Hemrajojha@123"
}
print(user_credentials)

# store a set of available items in a list 
items = ["samsung s25", "dell laptop", "Iphone 17"]
print(items)
my_cart = set()
```

```python
# prompt the user to enter a username and password, and validate the credentials against the dictionary
user_credentials = {
    "username" : "Hemraj123",
    "password" : "Hemrajojha@123"
}
user_name = input("Enter your name: ").lower()
pass_word = input("Enter your password: ")
if user_name == user_credentials['username'].lower() and pass_word == user_credentials['password']:
    print("You are logged in.")
    item = input("Enter an item you want to add in the cart: ")
    if item in items:
        my_cart.add(item)
        print(f"You have successfully added {item} into the cart.")
    else:
        print("Item not found.")
else:
    print("Credentials are not matchning.")
```
