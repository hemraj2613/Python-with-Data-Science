```python
# combining dataframes
```

```python
df1 = pd.DataFrame({'Col1':np.arange(5),'Col2':np.arange(5,10)})
df2 = pd.DataFrame({'Col1':np.arange(10,15),'Col2':np.arange(15,20)})
```

```python
pd.concat([df1,df2])
```

```python
df1 = pd.DataFrame({'Col1':np.arange(5),'Col2':np.arange(5,10)})
df2 = pd.DataFrame({'Col1':np.arange(10,15),'Col3':np.arange(15,20)})
```

```python
pd.concat([df1,df2],axis=1)
```

```python
pd.concat([df1,df2],axis=0)
```
