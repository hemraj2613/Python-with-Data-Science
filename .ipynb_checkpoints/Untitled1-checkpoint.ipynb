```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# pie chart

y = np.array([20,21,24,65,43])
plt.pie(y, )
plt.show()
```

```python
y = np.array([35,25,45,21])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
plt.pie(y, labels=mylabels, startangle=90)
plt.show()
```

```python
y =np.array([32,21,43,54,56,22])
mylabels = ["Apples", "Bananas", "Oranges", "Dates","Mangos", "Pineapples"]
mycolors = ["black", "hotpink", "b", "#4CAF50","r", "y"]

myexplode = [0.2,0,0,0,0,0]
plt.pie(y, labels=mylabels, explode=myexplode, colors=mycolors, shadow = True)
plt.legend(title="Six Fruits:")
plt.show()
```

```python
# histogram ==> it is a graph showing frequency distributions(in a given interval)

x = np.random.normal(100,10,20)
plt.hist(x)
plt.show()
```

```python
x = np.array(["India","Bhutan","Chine","Nepal"])
y = np.array([20,30,40,50])
plt.title("Countries vs Number",c="r" , size=23)
plt.xlabel("Countries")
plt.ylabel("Number")
plt.bar(x,y)
plt.show()
```

```python
x = np.array(["India","Bhutan","Chine","Nepal"])
y = np.array([20,30,40,50])
plt.title("Countries vs Number",c="r" , size=23)
plt.xlabel("Countries")
plt.ylabel("Number")
plt.bar(x,y, width=0.6, color="hotpink")
plt.show()
```

```python
x = np.array(["India","Bhutan","Chine","Nepal"])
y = np.array([20,30,40,50])
plt.title("Countries vs Number",c="r" , size=23)
plt.xlabel("Number")
plt.ylabel("Countries")
plt.barh(x,y, height=0.6, color="hotpink")
plt.show()
```

```python
x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
colors = np.array([0, 10, 20, 30, 40, 45, 50, 55, 60, 70, 80, 90, 100])
plt.scatter(x,y,c=colors,cmap='brg_r')
plt.colorbar()
plt.show()
```
