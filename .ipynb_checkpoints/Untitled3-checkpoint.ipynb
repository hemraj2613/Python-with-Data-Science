```python
# example of linear regression house price predictions
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Dataset
data = {
    'area_aana': [3, 4, 5, 6, 7, np.nan, 5, 8, 4, 5, 4, 50],
    'bedrooms': [2, 2, 3, 3, 4, 3, np.nan, 4, 2, 3, 2, 5],
    'age': [5, 4, 6, 3, 2, 5, 4, np.nan, 7, 5, 7, 1],
    'price_lakhs': [65, 75, 90, 110, 130, 100, 95, 145, 70, 90, 70, 900]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
# 2. Missing values
df.isnull().sum()
```

```python
df['area_aana'] = df['area_aana'].fillna(df['area_aana'].median())
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].median())
df['age'] = df['age'].fillna(df['age'].median())
```

```python
df
```

```python
# 3. Remove duplicates
# check duplicated
# df.duplicated().sum()
# df.duplicated()
df[df.duplicated()]
```

```python
df=df.drop_duplicates()
```

```python
df
```

```python
df.duplicated().sum()
```

```python
# 4. Outlier detection

# sort values
df['area_aana'].sort_values()
```

```python
Q1 = df['area_aana'].quantile(0.25)
Q3 = df['area_aana'].quantile(0.75)

IQR = Q3-Q1

lower_bound = Q1-1.5*IQR
upper_bound = Q3+1.5*IQR

outliers = df[
    (df['area_aana']<lower_bound)|(df['area_aana']>upper_bound)
]
```

```python
outliers
```

```python
lower_bound
```

```python
upper_bound
```

```python
# removes outliers
df = df[
    (df['area_aana'] >= lower_bound) &
    (df['area_aana'] <= upper_bound)
]
```

```python
df
```

```python
# 5. X and Y

X =df[['area_aana','bedrooms','age']]
Y =df['price_lakhs']
```

```python
# 6. Train-Test Split
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test= train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
# 7. Model

from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
# 8. Training
model.fit(X_train, Y_train)
```

```python
# 9. Prediction
Y_pred = model.predict(X_test)
```

```python
Y_pred
```

```python
# 10. Evaluation

from sklearn.metrics import (
mean_absolute_error, mean_squared_error, r2_score
)
MAE = mean_absolute_error(Y_test,Y_pred)
MSE = mean_squared_error(Y_test,Y_pred)
RMSE = np.sqrt(MSE)
R2_Score = r2_score(Y_test,Y_pred)
```

```python
print("MAE :", MAE )
print("MSE :", MSE )
print("RMSE :", RMSE )
print("R2_score :", R2_Score )
```
