```python
user_details = {
    "username":"Hem",
    "password":123}
class User:
    
    def __init__(self,username,password):
        self.__name = username
        self.__pass = password
        self.is_authenticated= False

    def auth(self):
        if self.__name == user_details['username'] and self.__pass == user_details['password']:
            self.is_authenticated = True
            print("User Logged in.")
        else:
            print("Invalid user.")

    def display_user(self):
        print(f"Logged in user is {self.__name}.\n")

class Blog:
    def __init__(self,title,author,content,user):
        self.blog_title = title
        self.blog_author = author
        self.blog_content = content
        self.user = user

    def show_blog(self):
        print(f"Title:{self.blog_title}")
        print(f"Author:{self.blog_author}")
        print(f"Content:{self.blog_content}")

class Comment:
    def __init__(self,text,blog):
        self.blog = None
        self.comment_text = text

    def show_comment(self):
        print(f"Comment: {self.comment_text}")


user = User("Hem",123)
user.auth()
if user.is_authenticated:
    blog = Blog(
        "Python",
        "Hemraj",
        "Object Oriented Programming.",
        user.display_user()
    )
    blog.show_blog()

    comment = Comment("This is a python blog!", blog.blog_title)
    comment.show_comment()
```
