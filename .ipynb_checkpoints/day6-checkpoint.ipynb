```python
# write a program to find whether a number is positive or negative
a=int(input("Enter first number:"))
if a>0:
      print("a is positive number")
else:
      print("a is negative number")  
```

```python
# elif
a=int(input("Enter first number:"))
if a>0:
      print("a is positive number")
elif a==0:
      print("a is zero")
else:
    print("a is negative number")
```

```python
a = 20
b= 30
if a>25 and b>25:
    print("Both are greater than 25")
elif a<25 and b<25:
    print("Both are less than 25")
elif a>25 or b>25:
    print("a or b one is greater")
else:
    print("Any one is less")
```

```python
# wap to check whether a number is greater or smaller 3 inputs
a=int(input("Enter a first number:"))
b=int(input("Enter a second number:"))
c=int(input("Enter a third number:"))
if a>b and a>c:
    print("a is greater")
elif b>a and b>c:
    print("b is greater")
else:
    print("c is greater")
```

```python
# nested if
if 20>10:
    if 20>15:
        if 20>18:
            print("20 is greater than 18")
        else:
            print("20 is less than 18")
    else:
        print("20 is less than 15")
else:
    print("20 is less than 10")
            
```
