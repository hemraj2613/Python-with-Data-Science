```python
# prime and not prime(composite)
numb = int(input("Enter a number: "))
number = [numb]
def prime(number):
    for num in number:
        if num ==2 or num == 3:
            print(f"{num} is prime")
        elif num<=1 or num % 2 == 0 or num % 3 == 0:
                print(f"{num} is not prime")
        else:
            print(f"{num} is prime")
prime(number)
```
