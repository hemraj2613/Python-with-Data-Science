```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as pl
```

```python
df = pd.read_csv('nepal_students_dataset.csv')
```

```python
df = df.head(10)
df
```

```python
# matplotlib
```

```python
# scatter

pl.figure(figsize=(7,5))
pl.title("Study hours vs Math marks")
pl.xlabel("Study Hours")
pl.ylabel("Marks obtained in Mathematics")
pl.scatter(df["study_hours_per_week"],df["math_marks"] , c='red', marker='*')
pl.grid()
pl.show()
```

```python
df = pd.read_csv('nepal_students_dataset.csv')
```

```python
df
```

```python
province_value_count = df['province'].value_counts()
```

```python
province_value_count
```

```python
province_value_count.values
```

```python
x =province_value_count.keys()
height = province_value_count.values
```

```python
height
```

```python
# bargraph
pl.figure(figsize=(10,4))
pl.title("Province name and data count")
pl.xlabel("Province Name")
pl.ylabel("Data Count")
pl.bar(x,height, label="Data Count", color="orange", edgecolor="black", alpha=0.5)
pl.legend()
pl.xticks(rotation=45)
# pl.yticks([0, 50, 100, 150, 200])
# pl.xlim(0, 5)
pl.show()
```
