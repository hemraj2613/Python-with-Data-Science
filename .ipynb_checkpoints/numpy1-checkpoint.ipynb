```python
# Create two 3×3 matrices:
# A = np.array([[2, 0, 1], [1, 3, 2], [4, 1, 5]])
# B = np.array([[1, 2, 0], [0, 1, 2], [2, 0, 1]])
# Compute the matrix product A · B using np.dot() or the @ operator (not element-wise multiplication).
# Compute the transpose of A and the determinant of A using np.linalg.det() .
# Check whether A is invertible (based on its determinant), and if so, compute its inverse using np.linalg.inv() .
# Verify your result by multiplying A with its inverse and confirming you get (approximately) the identity matrix.
```

```python
A = np.array([[2, 0, 1], [1, 3, 2], [4, 1, 5]])
B = np.array([[1, 2, 0], [0, 1, 2], [2, 0, 1]])
```

```python
np.dot(A,B)
```

```python
np.transpose(A)
```

```python
det_A = np.linalg.det(A)
det_A
```

```python
if not np.isclose(det_A, 0):
    print("\nA is invertible.")

else:
    print("\nA is not invertible.")
```

```python
A_inverse = np.linalg.inv(A)
A_inverse
```

```python
# Verify A × A⁻¹
identity = A @ A_inverse
identity
```

```python
print(np.allclose(identity, np.eye(3)))
```

```python
# You have a DataFrame df of employee records with columns ["EmpID", "Department", "Salary",
# "YearsExperience"] for 15 employees across 3 departments ( "Sales" , "IT" , "HR" ).
# Filter employees with YearsExperience > 5 and display only EmpID , Department , and Salary .
# Sort the full DataFrame first by Department (ascending) and then by Salary (descending).
# Use .apply() with a lambda function to create a new column "SeniorityLevel" : "Senior" if experience > 7,
# "Mid" if 3–7, otherwise "Junior" .
```

```python
data = {
    "EmpID": [101, 102, 103, 104, 105, 
              106, 107, 108, 109, 110,
              111, 112, 113, 114, 115],

    "Department": ["Sales", "IT", "HR", "Sales", "IT",
                   "HR", "Sales", "IT", "HR", "Sales",
                   "IT", "HR", "Sales", "IT", "HR"],

    "Salary": [45000, 65000, 50000, 55000, 75000,
               48000, 60000, 85000, 52000, 70000,
               90000, 58000, 62000, 80000, 55000],

    "YearsExperience": [2, 6, 8, 5, 10,
                        3, 7, 12, 1, 9,
                        15, 4, 6, 11, 2]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Filter employees with YearsExperience > 5 and display only EmpID , Department , and Salary .
filtered_exp = df[df["YearsExperience"]>5],["EmpID" , "Department" , "Salary"]
```

```python
filtered_exp
```

```python
# Sort the full DataFrame first by Department (ascending) and then by Salary (descending)
sorted_df = df.sort_values(
    by=["Department","Salary"],
    ascending=[True, False]
)
```

```python
sorted_df
```

```python
# Use .apply() with a lambda function to create a new column "SeniorityLevel" : "Senior" 
# if experience > 7, "Mid" if 3–7, otherwise "Junior" 
df["SeniorityLevel"] = df["YearsExperience"].apply(lambda x :"Senior" if x > 7 else "Mid" if x >= 3 else "Junior")
```

```python
df
```

```python
# A DataFrame df of 10 product records contains some missing values in the Price column and a few duplicate rows.
# Detect and count missing values per column.
# Fill missing Price values with the column's median.
# Identify and remove duplicate rows, keeping only the first occurrence.
```

```python
data = {
    "ProductID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer",
                "Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "Price": [80000, 1500, np.nan, 25000, 18000,
              80000, np.nan, 3500, 25000, 18000],
    "Quantity": [5, 20, 15, 8, 10, 5, 20, 12, 8, 10]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Detect and count missing values per column.
df.isnull().sum()
```

```python
# Fill missing Price values with the column's median.
median_price = df["Price"].median()
```

```python
df["Price"] = df["Price"].fillna(median_price)
```

```python
df
```

```python
# Identify and remove duplicate rows, keeping only the first occurrence
df.duplicated().sum()
```

```python
df = df.drop_duplicates(keep="first")
```

```python
# A retail company tracks daily temperature (°C) and ice-cream sales (units) for 30 days.
# Using np.random.seed(7) , generate temperature as 30 values drawn from np.random.normal(loc=25, scale=5, size=30) (round to 1 decimal), and sales as np.random.randint(50, 300, size=30) .
# Build a DataFrame with columns ["Day", "Temperature", "Sales"] where Day ranges from 1 to 30.
# Use np.where to create a new column "Weather" : "Hot" if Temperature > 28, "Mild" if between 18 and 28 (inclusive), otherwise "Cold" .
# Set a MultiIndex on ["Weather", "Day"] , then group by Weather to calculate the average Sales for each category.
# Compute the Pearson correlation coefficient between Temperature and Sales using np.corrcoef() , and briefly interpret whether hotter days are associated with higher sales.
```

```python
np.random.seed(7)
```

```python
data = np.random.normal(loc=25, scale=5, size=30)
```

```python
data
```

```python
np.round(data, 1)
```

```python
sales = np.random.randint(50, 300, size=30)
sales
```

```python
df = pd.DataFrame({
    "Day": range(1, 31),
    "Temperature": data,
    "Sales": sales
})
```

```python
df.head()
```

```python
df["Weather"] = np.where(df["Temperature"]>28,"Hot", np.where(
        df["Temperature"].between(18, 28),
        "Mild",
        "Cold"
    ))
```

```python
df.head()
```

```python
df = df.set_index(["Weather", "Day"])
```

```python
df
```

```python
# Group by Weather and calculate average Sales
avg_sales = df.groupby(level="Weather")["Sales"].mean()
```

```python
avg_sales
```

```python
# Pearson correlation coefficient
correlation_matrix = np.corrcoef(
    df["Temperature"],
    df["Sales"]
)
```

```python
correlation_matrix
```
