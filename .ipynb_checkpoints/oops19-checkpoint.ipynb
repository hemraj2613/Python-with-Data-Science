```python
# properties of OOP in python
# 1.Inheritance-->to hold properties of parent (by child)
# 2.Polymorphism-->
# 3.Encapsulation-->
# 4.Abstraction-->
```

```python
# inheritance

class Vehicle: #parent class
    wheel =100

    def drive(self):
        return "Hurray I am having fun driving."
#child class    
class Car(Vehicle): 
    engine = True

#child class
class Cycle(Vehicle): 
    gear = True
```

```python
car = Car()
```

```python
car.wheel
```

```python
cycle = Cycle()
```

```python
cycle.wheel
```

```python
car.drive()
```

```python
cycle.drive()
```

```python
car.engine
```

```python
cycle.gear
```

```python
# multiple inheritance
class Vehicle: #parent class
    wheel =100

    def drive(self):
        return "Hurray I am having fun driving."
#child class    
class Car(Vehicle): 
    engine = True

#child class
class Cycle(Car): 
    gear = True
```

```python
car1 = Car()
```

```python
car1.engine
```

```python
car1.engine
```

```python
cycle1 = Cycle()
```

```python
cycle1.engine
```

```python
cycle1.gear
```
