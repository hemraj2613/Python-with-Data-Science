```python
# sum of evens
values = [3,8,12,7,15,20,5]
sum = 0
for i in values:
    if i%2==0:
        sum = sum + i
print(sum)
```

```python
# sum of odd numbers 
sum = 0
for i in range(10):
    if i%2!=0:
        print(i)
        sum=sum+i
print("Sum is:",sum)
```

```python
# FizzBuzz-lite
numbers = [3,5,9,10,15,22,30]
for i in numbers:
    print(i)
    if i%3==0 and i%5==0:
        print("FizzBuzz")
    elif i%3==0:
        print("Fizz")
    elif i%5==0:
        print("Buzz")
    else:
        print("Nothing")
    print("\n")
```

```python
# nested loop table
x=[1,2,3,4,5]
for i in x:
    for j in x:
        print(f"{i} * {j} = {i * j}")
    print(" ")
```

```python
# product prices round
prices = {"apple": 1.5, "bread": 3.0, "milk": 2.25}
for i in prices.items():
    print(round(i[1],1))
```

```python
for item, value in prices.items():
    print(round(value,1))
```

```python
total = 0
for item, value in prices.items():
    print(f"{value:.2f}")
    total = total + value
print("Total is:", total)
    
```

```python
# predict the output
scores = [("Alice", 90), ("Bob", 72)]
for name, score in scores:
    if score >= 80:
        print(name, "passed with distinction")
    else:
        print(name, "passed")
```

```python
# Countdown
count = 5
while count>=1:
    print(count)
    count -= 1
print("Liftoff")
```

```python
# sum until threshold
values = [15,22,8,40,12,33,5]
length = len(values)
total = 0
index = 0
while index < length:
    total += values[index]
    index += 1
    if total >50:
        break
print("Total is ",total)
print(f"It took {index} values")
```

```python
# Skip Negatives
values = [4, -2, 9, -7, 3, -1, 8]
length = len(values)
total = 0
index = 0
while index < length:
    if values[index] < 0:
        index += 1
        continue
    total += values[index]
    index += 1
print("Total is:",total)
```

```python
# Retry Simulation 
attempts = iter(["wrong1", "wrong2", "correct"])
correct_password = "correct"

while True:
    attempt = next(attempts)

    if attempt != correct_password:
        print(" failed")
        continue

    print("successful")
    break
```

```python
# wap to find whether an item you are searching is present in the cart or not.
# yo neeed to take input from thr user(6 items)
# item you are trying to search, take an input from the user
# find whether item you are trying to search is present in cart or not
items = []
for i in range(6):
    item = input(f"Enter an item {i}")
    items.append(item)
```

```python
print(items)
```

```python
search = input("Enter an item to search")
```

```python
for item in items:
    if search == item:
        print(f"Item found {item}")
        break
else:
    print("Not found")
```
