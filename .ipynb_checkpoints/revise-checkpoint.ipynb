```python
# shallow copy cs deep copy
import copy
```

```python
# shallow copy
original = [[12,21,3,43],[99,90,87]]
shallow = copy.copy(original)
shallow[0].append(99)
print(original[0]) 
```

```python
shallow
```

```python
deep = copy.deepcopy(original)
deep
```

```python
#  *args, **kwargs
def func(*args, **kwargs):
    print(args)     
    print(kwargs)   

func(1, 2, x=3, y=4)
```
