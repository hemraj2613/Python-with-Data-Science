```python
# Outliers
# An outlier is a data value that is significantly different from the other values in a dataset.
```

```python
# get your dataset
# data = [10, 12, 11, 13, 15, 14, 100]
data = [21,30,27,18,25,27,26,24,20,22,21,60,70,80]
data.sort()
data
```

```python
import numpy as np
```

```python
# calculate Q1 and Q3
median = np.median(data)
Q1 = np.percentile(data,25)
Q3 = np.percentile(data,75)
```

```python
median
```

```python
Q1
```

```python
Q3
```

```python
# calculate IQR
IQR = Q3-Q1
```

```python
IQR
```

```python
# calculate the boundries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
lower_bound
```

```python
upper_bound
```

```python
outliers = [x for x in data if x < lower_bound or x > upper_bound]
```

```python
outliers
```
