```python
# Step 3: Make practice questions

words = ["user", "hi", "bot", "hello", "end"]
print(words[:1])
print(words[:3])
```

```python
words = chats[0].split()
for i in range(1, len(words)):
    print(words[:i], "->", words[i])
```

```python
inputs = [] # the words so far
answers = [] # the next word
for chat in chats:
    tokens = []
    for word in chat.split():
        tokens.append(word_to_token[word])
    for i in range(1, len(tokens)):
        inputs.append(tokens[:i])
        answers.append(tokens[i])
print(len(inputs), "practice questions")
print(inputs[3], "->", answers[3])
```

```python
inputs
```

```python
answers
```

```python
max_len = 9

X = keras.utils.pad_sequences(inputs, maxlen=9)
y = np.array(answers) # Keras wants the answers as a NumPy list too

print("Before:", inputs[3])
print("After :", X[3])
max_len
```

```python
X
```

```python
# Step 4: Build the model

vocab_size = len(word_to_token)
model = keras.Sequential([
layers.Embedding(vocab_size, 16),
layers.LSTM(64),
layers.Dense(vocab_size, activation="softmax"),
])

# model embedding, lstm (hit and trial with multiple number)
# "loss" measures the mistakes, "adam" is the tool that fixes them
model.compile(loss="sparse_categorical_crossentropy", optimizer="adam")
```

```python
# Step 5: Train the model

# batch_size=8: the model learns from 8 questions at a time

history = model.fit(X, y, epochs=300, batch_size=8, verbose=0)

# print("Loss at the start:", round(history.history["loss"][0], 2))
# print("Loss at the end :", round(history.history["loss"][-1], 2))
```

```python
# Step 6: Chat with the model

# "what is chatgpt chatgpt is a big model end"
text = "what is chatgpt chatgpt is big"
print("Input text: ",text)
tokens = []

for word in text.split():
    tokens.append(word_to_token[word])
    
print("Before Padding: ",tokens)

padded = keras.utils.pad_sequences([tokens], maxlen=max_len)
print("After Padding: ",padded)
[0,0,0,0,0,0,0,19,20]

chances = model.predict(padded, verbose=0)[0] # a chance for every word
best = chances.argmax() # token with the biggest chance

print("Predicted token",best)
print("Next word:", token_to_word[best])
print("FINAL: ",text+ " " +token_to_word[best])
```
