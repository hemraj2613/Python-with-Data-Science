```python
import gensim.downloader as api
glove = api.load("glove-wiki-gigaword-50")   # downloads once, then loads from disk

import numpy as np
import pandas as pd

from nltk.tokenize import word_tokenize
```

```python
data = pd.read_csv('Day1_Dataset.csv')
data.head()
```

```python
def review_to_vector(text):
    words = word_tokenize(text.lower())
    vectors = [glove[w] for w in words if w in glove]   # skip words GloVe doesn't know
    return np.mean(vectors, axis=0)
```

```python
review_to_vector(data["review"][0])
```

```python
# data['cleaned'] = data['review'].apply(preprocess)
```

```python
data['vectors'] = data['review'].apply(review_to_vector)
```
