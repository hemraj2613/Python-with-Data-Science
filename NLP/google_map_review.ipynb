```python
df = df[[
    "place_name",
    "place_category",
    "review_rating",
    "review_text",
    "review_date",
    "review_text_length"
]].copy()

df.head()
```

```python
def review_method(value):
    if value<3:
        return "Negative"
    elif value==3:
        return "Neutral"
    else:
        return "Positive"

review_method(1)
```

```python
df['sentiment'] = df['review_rating'].apply(review_method)
```

```python
# Data Preprocessing

import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer


# Create stopwords set
stop_words = set(stopwords.words('english'))

# Create the lemmatizer object once
lemmatizer = WordNetLemmatizer()

# Keep negation words because they affect sentiment
stop_words = stop_words - {"not", "no", "nor"}

def clean_text(text):
    text = text.lower()
    text = re.sub(r"<.*?>|http\S+|www\S+|@\w+|[^a-z\s]", "", text)
    words = word_tokenize(text)
    words = [
        word for word in words
        if word not in stopwords.words('english')
    ]
    words = [lemmatizer.lemmatize(lemmatizer.lemmatize(word, pos='v'), pos='n') for word in words]
    words = [word for word in words if len(word)>1]
    return " ".join(words)


# Apply preprocessing
df["cleaned_review"] = df["review_text"].apply(clean_text)

df
```
