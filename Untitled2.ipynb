```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
from sklearn.linear_model import LinearRegression
```

```python
steps:
1.Collect the data
2. visualize the data
3. Is it highly correlated?
4. Training(create object of class, train)
  4.1 Evaluate
  4.2 Validate
5. Prediction
```

```python
data = pd.DataFrame({
    'Qty': [1,2,3,5],
    'Amount': [80,160,240,400],
})
print("Mango qty in kg vs Amount in Rs")
data
```

```python
plt.scatter(data['Qty'], data['Amount'])
plt.grid()
plt.show()
```

```python
X = np.array(data['Qty'])
X= X.reshape(-1,1)
```

```python
y = np.array(data['Amount'])
y.reshape(-1,1)
```

```python
model = LinearRegression()
```

```python
model.fit(X, y)
```

```python
m =model.coef_
m
```

```python
c= model.intercept_
c
```

```python
m * 1 + c
```

```python
to_pridict = np.array(data['Qty']).reshape(-1,1)
```

```python
plt.scatter(data['Qty'], data['Amount'])
plt.plot(data['Qty'], model.predict(to_pridict))
plt.grid()
plt.show()
```

```python
df = pd.DataFrame({
    'House Size(in Aana)': [1,4,3,5,8,9],
    'House Price(in cr)': [2,6,4,8,12,14],
})
print("House Size vs House Price")
df
```

```python
plt.scatter(df['House Size(in Aana)'], df['House Price(in cr)'])
plt.title("House Size vs House Price")
plt.xlabel("House Size")
plt.ylabel("House Price(in Aana)")
plt.grid()
plt.show()
```

```python
X = np.array(df['House Size(in Aana)']).reshape(-1,1)
X
```

```python
y = np.array(df['House Price(in cr)']).reshape(-1,1)
y
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
m = model.coef_
m
```

```python
c = model.intercept_
c
```

```python
m * 7.5 + c
```

```python
plt.scatter(df['House Size(in Aana)'], df['House Price(in cr)'])
plt.plot(df['House Size(in Aana)'], model.predict(X))
plt.grid()
plt.show()
```
