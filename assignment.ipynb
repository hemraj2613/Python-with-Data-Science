```python
name = "Hem Raj Ojha"
print(name)
```

```python
# check if the string is starting with some text
print(name.startswith("Hem"))
```

```python
# check if the string is ending with some text
print(name.endswith("Ojha"))
print(name.endswith("12"))
```

```python
# check if the digit is in the string
text = "121swsw2"
print(text)
print(text.isdigit())
```

```python
text = "There are 2 books"
print(text)
has_digit = any(char.isdigit() for char in text)
print(has_digit)
```

```python
# removes white spaces from the string(beginning,ending, both)
name = "  python  "
print(name)
# beginning
print(name.lstrip())

# end
print(name.rstrip())

# both sides
print(name.strip())
```

```python
# how to capitalize the starting letter only
name = "hemraj ojha"
name.capitalize()
```

```python
# count the occurance of digits in string
text = "12python23"
count_digit = sum(char.isdigit() for char in text)
print(count_digit)
```

```python
# find the position of any digits inside the string
print(text)
positions = [index for index, char in enumerate(text) if char.isdigit()]
print(positions)
```

```python
# check whether a string is numeric
num = "123de42"
print(num.isnumeric())
```

```python
# check if the string is digit 
print(num.isdigit())
```

```python
# check if a string is alphanumeric
text = "hem12"
print(text.isalnum())
```

```python
# check if a string with decimal
num = "2332"
print(num.isdecimal())
```

```python
# examples related to string formatting
name = "Hemraj"
age = 23
print("My name is {} and I am {} years old.".format(name,age))
```
