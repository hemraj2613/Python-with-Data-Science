```python
# generator in python --> generates one at a time
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

gen = count_up_to(30)
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# dictionary comprehension
# {key:value for var in range}
{f"a{i}":i for i in range(5)}
```

```python
# lamda function--> short, anonymous, one line function
square = lambda x: x*x
print(square(6))
```

```python
sub = lambda a,b: a-b
print(sub(45,40))
```

```python
# check odd even
check = lambda x:"Even" if x%2==0 else "odd"
print(check(4))
```

```python
# wap to add a number into list using lambda function
l = [1,2,3,4]
n=4
add_list = lambda l,n: [i+n for i in l]
print(add_list(l,n))
print(add_list(n=n,l=l))
```

```python
# sum of squares
l = [1,2,3]
sum_n = 0
for i in l:
    sum_n = sum_n + i*i
print(sum_n)
```
