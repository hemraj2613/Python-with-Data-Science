```python
# Write a function that returns the smaller of two numbers
# if both are even, but returns the larger one if either number is odd.
def lesser_of_two_events(a,b):
    if a % 2 == 0 and b % 2 == 0:
        return min(a,b)
    else:
        return max(a,b)
```

```python
lesser_of_two_events(3,4)
```

```python
lesser_of_two_events(2,4)
```

```python
# Animal crackers
# Write a function that takes a two-word string and returns 
# True if both words start with the same letter.
def animal_crackers(text):
    words = text.split()
    return words[0][0] ==words[1][0]
animal_crackers("Harry Hari")
```

```python
# MAKES TWENTY
# Given two integers, return True if their 
# sum is 20, or if either one of them is 20 on its own.

def makes_twenty(n1,n2):
    return (n1 + n2) == 20 or n1==20 or n2==20

makes_twenty(30,20)
```

```python
# Write a function that capitalizes 
# both the first and the fourth letter of a name.
def capital_word(name):
    if len(name) > 3:
        return name[:3].capitalize() + name[3:].capitalize()
    else:
        return "Name is too short"

capital_word("hemraj")
```

```python
# Given a sentence, return it with the words in reverse order.
def reverse_word(name):
    return " ".join(name.split()[::-1])
```

```python
reverse_word("My name is Hem")
```

```python
# Given an integer n , return True if it's within 10 of either 100 or 200.
def almost_there(n):
    return (abs(100-n)<=10 or abs(200-n)<=10)
```

```python
almost_there(20)
```

```python
almost_there(110)
```

```python
# Given a list of integers, return True if it 
# contains two 3's sitting right next to each other.
def has_33(nums):
    for i in range(0, len(nums)-1):
        if nums[i] == 3 and nums[i+1] == 3:
            return True
    return False

has_33([1,2,3,3])
```

```python
# Given a string, return a new string 
# where every character is repeated three times in a row.
def paper_doll(text):
    result = ""
    for char in text:
        result += char * 3
    return result
paper_doll("Hello")
```
