```python
user_details = {
    "username":"Hem",
    "password":123}
class User:
    
    def __init__(self,username,password):
        self.__name = username
        self.__pass = password
        self.is_authenticated= False

    def auth(self):
        if self.__name == user_details['username'] and self.__pass == user_details['password']:
            self.is_authenticated = True
            print("User Logged in.")
        else:
            print("Invalid user.")

    def display_user(self):
        print(f"Logged in user is {self.__name}.\n")

class Blog:
    def __init__(self,title,author,content,user):
        self.blog_title = title
        self.blog_author = author
        self.blog_content = content
        self.user = user

    def show_blog(self):
        print(f"Title:{self.blog_title}")
        print(f"Author:{self.blog_author}")
        print(f"Content:{self.blog_content}")

class Comment:
    def __init__(self,text,blog):
        self.blog = None
        self.comment_text = text

    def show_comment(self):
        print(f"Comment: {self.comment_text}")


user = User("Hem",123)
user.auth()
if user.is_authenticated:
    blog = Blog(
        "Python",
        "Hemraj",
        "Object Oriented Programming.",
        user.display_user()
    )
    blog.show_blog()

    comment = Comment("This is a python blog!", blog.blog_title)
    comment.show_comment()
```

```python
class User:
    def __init__(self,username,password,email,phone):
        self.__name = username
        self.__pass = password
        self.email = email
        self.phone = phone
        self.is_authenticated= False

    def auth(self,username,password):
        if username == self.__name and password == self.__pass:
            self.is_authenticated = True
            print("User Logged in.")
        else:
            print("Invalid user.")

    def display_user(self):
        return self.email + self.phone

class Blog:
    def __init__(self,title,author,content):
        self.blog_title = title
        self.blog_author = author
        self.blog_content = content

    def show_blog(self):
        print(f"Title:{self.blog_title}")
        print(f"Author:{self.blog_author.display_user()}")
        print(f"Content:{self.blog_content}")

class Comment:
    def __init__(self,text,blog,user):
        self.user = user
        self.blog = blog
        self.comment_text = text

    def show_comment(self):
        print(f"Comment: {self.comment_text}")
```

```python
class InsufficientFundsError(Exception):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(f"Tried to withdraw {amount}, but balance is only {balance}")
    return balance - amount
# withdraw(100, 250)

try:
    withdraw(100, 250)
except InsufficientFundsError as e:
    print("InsufficientFundsError:", e)
```

```python
# Write a function safe_divide(a, b) that returns a / b , but catches 
# ZeroDivisionError and prints a warning message (instead of crashing) when b is 0 .

def safe_divide(a,b):
    return a/b

try:
    safe_divide(a=4,b=0)
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)
```

```python
# Write a function safe_get(items, index) that returns items[index] , but catches 
# IndexError and prints a warning message (instead of crashing) when the index is out of range.

def safe_get(items, index):
    return items[index]

try:
    items = [2,4,6,7,2]
    items[6]
except IndexError as e:
    print("IndexError:", e)
```

```python
# Write a function set_temperature(celsius) that raise s a ValueError if celsius is below -273.15 (absolute zero), 
# and otherwise returns the value. Test it with both a valid and an invalid temperature, using try / except to catch the invalid one.

def set_temperature(celcius):
    if celcius < -273.15:
        raise ValueError("Temperature cannot be less than -273.15.")
    return celcius
# set_temperature(-175.21)

try:
    set_temperature(-175.21)
    print("Valid temperature")
except ValueError as e:
    print("ValueError: ", e)
```

```python
# Write a custom exception NegativeAgeError (subclassing Exception ) and a function register_user(name, age) 
# that raises it if age is negative. Catch it with try / except and print a friendly message.
class NegativeAgeError(Exception):
    pass
    
def register_user(name, age):
    if age < 0:
        raise NegativeAgeError("Age cannot be negative.")
    return age
# register_user("Hem", -21)

try:
    register_user("Hem", -21)
except NegativeAgeError as e:
    print("NegativeAgeError:", e)
```
