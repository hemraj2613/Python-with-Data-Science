```python
# wap to (.txt file)
# 1. open a file ('broadway.txt') -->try
# 2.read the contents of this file --> try
# 3. if file doesn't exist, create a file --> except
# 4. add some data into it and create a file --> except
# 5. read the content again --> finally
```

```python
try:
    f1 = open("broadway.txt", 'r')
    print(f1.read())
except FileNotFoundError:
    f1 = open("broadway.txt", 'w')
    f1.write("Welcome to my file.")
    f1.close()
finally:
    f1 = open("broadway.txt", "r")
    print("Final file content:")
    print(f1.read())
    f1.close()
```

```python
data = pd.DataFrame({
    "Age":[35,36,38,44,59],
    "Salary":[70,55,80,44,30],
    "Location":["Banesor","Kalanki","Putalisadak","Tinkune","Kotesor"]
})
```

```python
data
```

```python
data['Age']
```

```python
pl.title("Age vs Salary")
pl.xlabel("Age")
pl.ylabel("Salary")
pl.scatter(data["Age"], data["Salary"])
pl.grid()
pl.show()
```

```python
pl.title("Age vs Salary")
pl.xlabel("Age")
pl.ylabel("Salary")
pl.plot(data["Age"], data["Salary"],marker = 'o', ms = 20, mfc = 'r')
pl.grid()
pl.show()
```

```python
pl.title("Location vs Age")
pl.xlabel("Location name")
pl.ylabel("Age Value")
pl.bar(data["Location"], data["Age"])
pl.show()
```

```python
pl.title("Location vs Salary")
pl.xlabel("Location name")
pl.ylabel("Salary Value")
pl.bar(data["Location"], data["Salary"])
pl.show()
```

```python
pl.title("Location vs Salary")
pl.pie(data["Salary"], labels=data["Location"], autopct="%0.2f%%")
pl.show()
```

```python
pl.title("Age vs Salary")
pl.xlabel("Salary")
pl.ylabel("Age")
pl.hist(data["Salary"], data["Age"], color='skyblue', edgecolor='black')
pl.show()
```
