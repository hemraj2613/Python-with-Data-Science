```python
# wap to create list of your friends inside a dictionary of each class
# take input to search for which friend you want to search
# check if your friend is in your friends list or not and print which class he belongs to
```

```python
friends = {
    "class1":["Mohan","Sony"],
    "class2":["Hari","Nabin"],
}
print(friends)
```

```python
friend_name= input("Enter your friend name: ")
print("Friend name is:", friend_name)
found = False
for i in friends.items():
    inner_list = i[1]
    if friend_name in inner_list:
        print(f"Your friend is belongs to {i[0]}")
        found = True
        break
if not found:
    print("Not present")
```
