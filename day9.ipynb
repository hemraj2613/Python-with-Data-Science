```python
l1 = []
l2 = []
num = int(input("Enter a number: "))
for i in range(num):
    if i % 2 == 0:
        l1.append(i)
    else:
        l2.append(i)
print("Even number:",l1)
print("Odd number:",l2)
```

```python
# nested for loop
l1 = [1,2]
l2 = [3,4]
for i in l1:
    for j in l2:
        print(i)
        print(j)
```
