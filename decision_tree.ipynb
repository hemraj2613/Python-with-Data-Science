```python
import pandas as pd
import matplotlib.pyplot as pl
```

```python
data = pd.DataFrame(
    {
        'Loves Momo':[1,1,0,0,1,1,0],
        'Loves Pepsi':[1,0,1,1,1,0,0],
        'Loves Thakali':[0,0,1,1,1,0,0],
    }
)
```

```python
data
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier()
```

```python
X = data[['Loves Momo','Loves Pepsi']]
y = data['Loves Thakali']
```

```python
X
```

```python
y
```

```python
model.fit(X,y)
```

```python
from sklearn.tree import plot_tree
```

```python
pl.figure(figsize=(20,8))
plot_tree(
    model,
    feature_names=X.columns,
    class_names=["Loves Thakali", "Doesnt loves Thakali"],
    filled=True,
    fontsize=18,
)
pl.show()
```

```python
model.predict([[1,0]])
```
