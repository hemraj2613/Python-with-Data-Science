```python
1. What is Deep Learning?
Deep Learning is a subset of Machine Learning that uses 
Artificial Neural Networks (ANNs) with multiple layers to 
learn complex patterns from data.
```

```python
Example
Suppose we want to identify whether an image contains a cat or dog.
```

```python
Traditional ML:
Image → Feature Extraction → ML Algorithm → Cat/Dog
```

```python
Deep Learning:
Image → Neural Network → Cat/Dog
```

```python
Deep Learning automatically learns useful features from the data.
```

```python
2. Artificial Neural Network (ANN)
ANN is inspired by the human brain.

A neural network consists mainly of:
Input Layer
Hidden Layer(s)
Output Layer
```

```python
3. Neuron
A neuron receives inputs, multiplies them by weights, 
adds bias, and passes the result through an activation function.
```

```python
Formula:
z=w1*x1+w2*x2+...+wn*xn+b

Then:
y=f(z)
```

```python
Where:

x = input
w = weight
b = bias
z = weighted sum
f = activation function
y = output
```

```python
5. Perceptron
A Perceptron is one of the simplest neural network models.

It is mainly used for binary classification.
```
