```python
import numpy as np
```

```python
w1 = np.random.randint(-10,10)
w2 = np.random.randint(-10,10)
b = np.random.randint(-10,10)
```

```python
z1 = 0*w1 + 0*w2 + b
z2 = 0*w1 + 1*w2 + b
z3 = 1*w1 + 0*w2 + b
z4 = 1*w1 + 1*w2 + b
```

```python
z1,z2,z3,z4
```

```python
def activation_func(z):
    if z>0:
        return 1
    else:
        return 0
```

```python
f_z1 = activation_func(z1)
f_z2 = activation_func(z2)
f_z3 = activation_func(z3)
f_z4 = activation_func(z4)
```

```python
f_z1, f_z2, f_z3, f_z4
```

```python
if f_z1==0 and f_z2==1 and f_z3==1 and f_z4==1:
    print("Actual and predicted matched, so w1, w2 are final values")
else:
    print("Error occured")
```

```python
x1=(0,0,1,1)
x2=(0,1,0,1)
out=(0,1,1,1)
for i in range(10):
    print(f"\n------Index{i} -------")
    w1 = np.random.randint(-10,10)
    w2 = np.random.randint(-10,10)
    b = np.random.randint(-10,10)

    print(f"Weight1: {w1}, Weight2: {w2}, Bias: {b}")

    z1 = 0*w1 + 0*w2 + b
    z2 = 0*w1 + 1*w2 + b
    z3 = 1*w1 + 0*w2 + b
    z4 = 1*w1 + 1*w2 + b

    def activation_func(z):
        if z>0:
            return 1
        else:
            return 0

    f_z1 = activation_func(z1)
    f_z2 = activation_func(z2)
    f_z3 = activation_func(z3)
    f_z4 = activation_func(z4)

    print(f"f_z1: {f_z1}, f_z2: {f_z2}, f_z3: {f_z3}, f_z4: {f_z4}")
    
    if f_z1==0 and f_z2==1 and f_z3==1 and f_z4==1:
        print("Actual and predicted matched, so w1, w2 are final values")
        
    else:
        print("Error occured")
```
