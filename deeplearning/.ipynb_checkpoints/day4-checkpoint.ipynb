```python
# pip install tensorflow matplotlib   (if needed)
import numpy as np
import matplotlib.pyplot as plt
import keras
from keras import layers
 
# float64 so the tiny late-iteration numbers match the slides
# (the default float32 is too coarse to show values like 1e-9)
 
x = np.array([1, 2, 4, 5, 6], dtype="float64").reshape(-1, 1)   # size (Ropani)
y = np.array([11, 20, 38, 47, 56], dtype="float64").reshape(-1, 1)  # price (Crores)
 
print(np.hstack([x, y]))
```

```python
x
```

```python
y
```

```python
model = keras.Sequential([
    keras.Input(shape=(1,)),
    layers.Dense(
        1,  #neurons = 1
        kernel_initializer="zeros",  #w=0
        bias_initializer="zeros"  #b=0
    ),
])
```

```python
model
```

```python
model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.03),  # plain gradient descent, alpha = 0.03
    loss="mse"   #(1/n) * sum((y_hat-y)^2)
)

model.summary()
```

```python
history = model.fit(
    x, y,
    epochs=5,
    batch_size=5,   #all 5 rows in one batch -> true gradient descent
    shuffle=False,
    verbose=1,
)
```

```python
w, b = model.get_weights()
print(f"Learned: w={w[0,0]:.6f}, b={b[0]:.6f}")
```
