```python
import os
from PIL import Image
import numpy as np
import matplotlib.pyplot as pl
import cv2
from tensorflow import keras
from sklearn.model_selection import train_test_split
```

```python
# one way to load images
```

```python
with_mask = os.listdir("data/with_mask/")
print(with_mask[0:5])
```

```python
without_mask = os.listdir("data/without_mask/")
print(without_mask[0:5])
```

```python
print("Number of mask images:", len(with_mask))
print("Number of without mask images:", len(without_mask))
```

```python
# creating labels for the two images 

# with_mask ==> 1
# without_mask ==> 0
```

```python
# create the labels

with_mask_labels = [1]*len(with_mask)
without_mask_labels = [0]*len(without_mask)
```

```python
print("With_mask labels:", with_mask_labels[0:5])
print("Without_mask labels:", without_mask_labels[0:5])
```

```python
# combine both labels

labels =  without_mask_labels + without_mask_labels

print(len(labels))
```
