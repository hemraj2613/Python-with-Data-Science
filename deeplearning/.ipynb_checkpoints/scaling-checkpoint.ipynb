```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({'Col1':[10,20,22,40,50],

                   'Col2':[55,66,44,88,99]})

df
 
```

```python
plt.scatter(df['Col1'],df['Col2'])
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
df2 = pd.DataFrame(scaler.fit_transform(df), columns=['Col1','Col2'])
```

```python
df2
```

```python
plt.scatter(df2['Col1'],df['Col2'])
```
