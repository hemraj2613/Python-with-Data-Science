```python
Write a python program...
Do this 100 times inside a loop
    step 1: Initialize weights(w1) randomly
    atep 2: Initialize the bias(b) randomly
    step 3: Compute: z = w1 * x +  b
    step 4: Activaton funcction: --> f(z) = z no transfomation and activation function required
    step 5: Compare predicted vs actual all(actual-predicted)
    step 6: until zero errors are found goto step 1
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Size": [4, 6, 8, 10, 12],
    "Price": [85, 115, 145, 175, 205]
})

x = df["Size"].values
y = df["Price"].values

for i in range(5):
    print(f"\n------ Index {i} ------")
    w1 = np.random.randint(-20, 30)
    b = np.random.randint(-20, 40)
    
    print(f"Weight1: {w1}, Bias: {b}")
    total_error = 0

    for j in range(len(x)):
        z = w1 * x[j] + b
        f_z = z

        # Compare actual vs predicted
        error = y[j] - f_z

        print(
            f"Size={x[j]}, Actual={y[j]}, Predicted={f_z}, Error={error}"
        )

        total_error += abs(error)

    print(f"Total Error: {total_error}")

    if total_error == 0:
        print("Actual and predicted matched!")
        print(f"Final Weight1 = {w1}")
        print(f"Final Bias = {b}")
        break

    else:
        print("Error occurred")
```
