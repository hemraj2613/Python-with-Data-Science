```python
image = cv2.imread("./images/astronaut.png")

plt.imshow(image)
```

```python
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
# image flip

image_flip = cv2.flip(image,0)
plt.imshow(image_flip)
plt.show()
```
