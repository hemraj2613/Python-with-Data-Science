```python
# Conceptual Understanding
1. In one sentence, what problem does Linear Regression solve?
=> It solves continuous numerical value problems.
```

```python
2. What's the difference between a "continuous" output and a 
"categorical" output? Give one example of each.
=> Continuous output refers to the numerical value (i.e. whole and decimal numbers)
eg: House Price = Rs. 5cr
=> Categorical output refers to the category value
eg: Email: Spam /Not Spam
```

```python
3. Is predicting whether an email is "spam" or "not spam" 
a regression problem or a classification problem? Why?
=> It is Classification problem. Because, it is classified into 
two options(category) not in numerical value.
```

```python
4. Why can't we just rely on guesswork instead of 
using a model like Linear Regression?
=> Always guessing doesnot make the best fit decisions but model uses the 
patterns and data to make predictions.
```

```python
# Terminology
5. In the equation y = mx + b, what do each of 
the four symbols (x, y, m, b) represent?
=> x = input  variable
y = output variable
m = slope
b = intercept
```

```python
6. What is the difference between an independent variable 
and a dependent variable? Give an example from the house price dataset.
=> Independent variable(x): input value used for predictions (eg: House size)
=> Dependent variable(y) : Output value that we want to predict (eg: House price)
```

```python
7. What does the "coefficient" (or slope) of a line tell us?
=> It tells how much the output value changes when the input increased
```

```python
8. What does the "intercept" of a line represent?
=> The intercept is the predicted value of y when x = 0
```

```python
9. What is a "residual," and how do you calculate it?
=> It is the difference between the actual value and the predicted value.
```

```python
# Applying the Concept
10. If the best-fit line is price = 0.15 × size + 30, 
what price would this model predict for a house that is 1,200 sq. ft.?
=> price = 0.15 x 1200 + 30 = 210
```

```python
11. A model predicts a house will sell for $220K, but it actually sold for $235K. 
What is the residual, and does this mean the model overestimated or underestimated the price?
=>residual = actual price - predicted price 
  residual = 235-220 = $15
=> Model underestimated by $15 (because theactual price is higher than the predicted price)
```

```python
12. Why is predicting a value within the range of your original data 
(interpolation) generally safer than predicting far outside that range (extrapolation)?
=> Interpolation is generally safer because the model has already seen data from that 
range and can identify the relationship more reliably.

=> Extrapolation is riskier because the relationship may change outside the 
range of the original data.
```

```python
# Best-Fit Line & Residuals
13. What makes a line the "best-fit" line among all possible 
lines you could draw through a set of points?
=> Less/Small error/residuals between the actual and predicted values makes a line "best-fit" 
```

```python
14. Why do we square the residuals instead of just adding them up directly?
=> 1. Positive and negative residuals could cancel each other.
=> 2. Squaring makes all errors positive and gives larger errors more importance.
```

```python
# Bigger Picture
15. Why is this method called "Linear" Regression — what assumption 
is being made about the relationship between the input and output?
=> This method is called "Linear" regression because it tells the 
relationships between input and the output variable 
and represented by the straight line.
```
