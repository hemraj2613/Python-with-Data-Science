```python
# check if a number is prime
def is_prime(n):
    if n < 2:
        return f"Number {n} is not prime"
    for i in range(2,n):
        if n % i == 0:
            return f"Number {n} is not prime"
    return f"Number {n} is prime"
```

```python
is_prime(4)
```

```python
# The employee_of_the_month function returns both 
# the name and hours of whoever worked the most.

work_hours = [("Abby", 100), ("Billy", 400), ("Cassie", 800)]
def emp_month(work_hours):
    max_hrs = 0
    top_emp = ""
    for emp, hours in work_hours:
        if hours > max_hrs:
            max_hrs = hours
            top_emp = emp
        else:
            pass
    return (top_emp, max_hrs)

emp_month(work_hours)
```

```python
# gussing game
def player_guess():
    guess = ""
    simulated_inputs = iter(["not a number", "31", "1"])

    while guess not in ["0", "1", "2"]:
        guess = next(simulated_inputs)
        print(f"Pick a number: 0, 1 or 2: {guess}")

    return int(guess)

player_guess()
```

```python
# sum of squares
# Write a function sum_of_squares(numbers) that returns 
# the sum of the squares of every number in a list. 
# Test it on [1, 2, 3, 4] (should return 30 ).

num_list = [1, 2, 3, 4]
sqr = []
def sum_of_squares(num_list):
    add_sq = 0
    for i in num_list:
        i = i*i
        sqr.append(i)
        add_sq += i
    return sqr, add_sq
    print("\nThe sum of squares is :",add_sq)
sum_of_squares(num_list)
```

```python
# Find the Longest Word (Medium)
# Write a function longest_word(sentence) 
# that takes a string, splits it into words, 
# and returns the longest one. Test it on "the quickbrown fox jumps" .

def longest_word(sentence):
    words = sentence.split()
    longest = words[0]

    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest

sentence = "the quickbrown fox jumps"
longest_word(sentence)
```

```python
# Min and Max Together (Medium)
# Write a function min_and_max(numbers) that returns a tuple of 
# (smallest, largest) from a list, then unpack the result into two
# separate variables when you call it. Test it on [4, 9, 1, 7, 3] .

def min_and_max(numbers):
    smallest = min(numbers)
    largest = max(numbers)

    return (smallest, largest)

numbers = [4, 9, 1, 7, 3]
small,large = min_and_max(numbers)

print(f"Smallest: {small}")
print(f"Largest: {large}")
```

```python
# A Function That Calls Another (Medium/Hard)
# Write two functions: is_even(n) returning a boolean, 
# and count_evens(numbers) that calls is_even() internally to count how
# many even numbers are in a list. 
# Test count_evens([1, 2, 3, 4, 5, 6]) (should return 3 ).

def is_even(n):
    even_num = []
    
    for num in n:
        if num % 2 == 0:
            even_num.append(num)
    return even_num
    
n = [1,2,3,4,5,6]
result = is_even(n)
print("Even number list:", result)
```

```python
def count_even(n):
    even_count = is_even(n)
    return len(even_count)

print("Count of even numbers are:",count_even(n))
```
