```python
# You have two DataFrames:

# Merge orders and customers on CustomerID using an inner join.
# Create a pivot table showing total Amount spent per City .
# Identify the customer with the highest total spending and print their name and city.
```

```python
orders = pd.DataFrame({
    "OrderID": [1, 2, 3, 4, 5],
    "CustomerID": [101, 102, 101, 103, 102],
    "Amount": [250, 450, 300, 600, 150]
})
customers = pd.DataFrame({
    "CustomerID": [101, 102, 103],
    "Name": ["Sita", "Ramesh", "Gita"],
    "City": ["Kathmandu", "Pokhara", "Lalitpur"]
})
```

```python
df= pd.merge(
    orders,
    customers,
    on='CustomerID',
    how='inner'
)
```

```python
df
```

```python
pd.pivot_table(df,values='Amount', index='City', aggfunc='sum')
```

```python
df
```

```python
df.groupby(['Name','City'])['Amount'].sum()
```
