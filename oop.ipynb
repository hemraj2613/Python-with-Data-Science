```python
# class
class Car:
    def color(self,c):
        return f"The car is {c}."
    def size(self,size_car):
        return f"The size is {size_car}."
    def speed(self,speed_car):
        return f"The speed of car is {speed_car}."
    def wheel(self,wheel_num):
        return f"The number of wheels are {wheel_num}."
```

```python
# object1
car1 = Car()
print(car1.color("red"))
print(car1.size("heavy"))
print(car1.speed("100km/hr"))
print(car1.wheel("4"))
```

```python
# object2
car2 = Car()
print(car2.color("Blue"))
print(car2.size("small"))
print(car2.speed("90km/hr"))
print(car2.wheel("4"))
```

```python
# __init__() is used to initialized attributes
# attributes == variables of class
# set_roll_number() is a method os a class

class RollNumber:
    def __init__(self):
        self.roll_number = None
    def set_roll_number(self,roll_number):
        self.roll_number = roll_number
    def get_roll_number(self):
        return self.roll_number
```

```python
obj1 = RollNumber()
print(obj1.roll_number)
print(obj1.set_roll_number(2))
```

```python
print(obj1.get_roll_number())
```

```python
# calculator using class and object

class Calculator:
    def add(self,a,b):
        return f"The operation is addition: {a+b}."
    def sub(self,a,b):
        return f"The operation is subtraction: {a-b}."
    def mul(self,a,b):
        return f"The operation is multiplication: {a*b}"
    def div(self,a,b):
        return f"The operation is division: {a/b}"
```

```python
calc = Calculator()
print(calc.add(2,3))
print(calc.sub(12,5))
print(calc.mul(22,3))
print(calc.div(32,2))
```
