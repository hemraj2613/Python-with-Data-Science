```python
# inheritance

class Vehicle: #parent class
    wheel =100

    def drive(self):
        return "Hurray I am having fun driving."
#child class    
class Car(Vehicle): 
    engine = True

#child class
class Cycle(Vehicle): 
    gear = True
```

```python
car = Car()
```

```python
car.wheel
```

```python
cycle = Cycle()
```

```python
cycle.wheel
```

```python
car.drive()
```

```python
cycle.drive()
```

```python
car.engine
```

```python
cycle.gear
```

```python
# multiple inheritance
class Vehicle: #parent class
    wheel =100

    def drive(self):
        return "Hurray I am having fun driving."
#child class    
class Car(Vehicle): 
    engine = True

#child class
class Cycle(Car): 
    gear = True
```

```python
car1 = Car()
```

```python
car1.engine
```

```python
car1.engine
```

```python
cycle1 = Cycle()
```

```python
cycle1.engine
```

```python
cycle1.gear
```

```python
# Create a class Rickshaw with driver_name and fare_per_km. 
# Make two objects and print their details.

class Rickshaw:
    def __init__(self,driver_name,fare_per_km):
        self.d_name = driver_name
        self.f_km = fare_per_km
    def print_details(self):
        print(f"Driver name is {self.d_name}.")
        print(f"Fare per km: {self.f_km}")
        
rickshaw1 = Rickshaw(driver_name="Hem", fare_per_km=230)
rickshaw1.print_details()
```

```python
rickshaw2 = Rickshaw(driver_name="Rajan", fare_per_km=260)
rickshaw2.print_details()
```

```python
# Create a class Shop with shop_name and owner_name. 
# Add a method greet() that prints a welcome message.

class Shop:
    def __init__(self,shop_name,owner_name):
        self.s_name = shop_name
        self.o_name = owner_name
    def greet(self):
        print(f"Welcome {self.o_name} owner of {self.s_name}.")
shop1 = Shop(shop_name="HR Electronic",owner_name="Hemraj")
shop1.greet()
```

```python
# Create a class Farmer with name and land_area
# (in ropani). Add a method describe().

class Farmer:
    def __init__(self,name,land_area):
        self.f_name = name
        self.l_area = land_area
    def describe(self):
        print(f"{self.f_name} is a farmer and has {self.l_area} land area.")

farmer = Farmer(name="Dammar",land_area="12 ropani")
farmer.describe()
```

```python
# Create a class MobileRecharge with customer_name 
# and balance. Add a method recharge(amount).
class MobileRecharge:
    def __init__(self,customer_name,balance):
        self.customer_name = customer_name
        self.balance = balance
    def __str__(self):
        return (f"Dear {self.customer_name} your current balance is {self.balance}.")
    def recharge(self,amount):
        print(f"Dear {self.customer_name} your recharge amount is {amount}.")
        amount += self.balance
        print(f"Dear {self.customer_name} your new balance after recharge amount is {amount}.")
mobile_recharge = MobileRecharge(customer_name="Hemraj", balance=10)
print(mobile_recharge)
```

```python
mobile_recharge.recharge(amount=49.0)
```

```python
# Create a class Resturant with name and location. 
# Make three objects and print each one's details.

class Resturant:
    def __init__(self,name,location):
        self.name = name
        self.location = location
    def details(self):
        print(f"{self.name} resturant is located at {self.location}.")
r1 = Resturant(name="3D Resturant", location="Lalitpur")
r2 = Resturant(name="Java Resturant", location="Baneshwor")
r3 = Resturant(name="DaiVai Resturant", location="Bhaktspur")

r1.details()
r2.details()
r3.details()
```

```python
# Create a class Book with attributes title, author, 
# and price, set through a constructor. Add a method 
# display_details() that prints all three.
# Create two Book objects with different values and call display_details() on each.

class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    def display_details(self):
        print(f"The title is {self.title}.")
        print(f"The title is {self.author}.")
        print(f"The title is {self.price}.")
b1 = Book(title="Atomic Habit", author="James", price="500")
b2 = Book(title="Muna Madan", author="Devkota", price="300")
print(b1.display_details())
print(b2.display_details())
```

```python
class Taxi:
    def __init__(self, plate_number, fare_per_km):
        self.plate_number = plate_number
        self.fare_per_km = fare_per_km
    def calculate_fare(self, distance):
        total = self.fare_per_km * distance
        print(f"Total fare is {total}")
        
t1 = Taxi(plate_number=201, fare_per_km=300)
t1.calculate_fare(12)
```

```python
class Employee:
    def __init__(self, name, basic_salary):
        self.name = name
        self.basic_salary = basic_salary
    def calculate_salary(self):
        bonus = 0.20 * self.basic_salary
        total = self.basic_salary + bonus
        print(f"Employee name is {self.name} and total salary is {total}.")
e1 = Employee(name = "Hemraj", basic_salary =20000)
e2= Employee(name = "Dammar", basic_salary =30000)

e1.calculate_salary()
```

```python
e2.calculate_salary()
```

```python
# inheritance

class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def show_details(self):
        print(f"Vehicle brand is {self.brand} and its speed is {self.speed}.")
        
class Bike(Vehicle): # inherit from Vehicle
    def show_type(self):
        return "This is a two-wheeler."
    
bike1 = Bike(brand="Maruti", speed="100km/hr")
bike1.show_details()
bike1.show_type()
```
