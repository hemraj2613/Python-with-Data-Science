```python
# Create a new list containing only the larger element at each index.
l1 = [10, 25, 30, 15, 50]
l2 = [20, 15, 35, 10, 40]
```

```python
l3=[]
for i in range(len(l1)):
    l3.append(max(l1[i],l2[i]))
print(l3)
```

```python
# Create a new list containing the cube of every element.
l = [2,3,4,5]
```

```python
l1 = []
for i in range(len(l)):
    l1.append(l[i]**2)
print(l1)
```

```python
# Add 10 to every element of a list.
l = [2,3,4,5]
```

```python
l1=[]
for i in range(len(l)):
    l1.append(l[i]+10)
print(l1)
```

```python
# Find the sum of all elements without using sum().
l = [2,3,4,5]
```

```python
sumNum =0
for i in range(len(l)):
    sumNum = sumNum + l[i]
print(sumNum)
```

```python
# Find the largest element without using max().
l = [12,3,44,5]
lar = l[0]
for i in range(len(l)):
    for j in range(i):
        if l[i]>lar:
            lar = l[i]
print(lar)
```

```python
# Reverse a list without using reverse() or slicing [::-1].
l = [1, 2, 3, 4, 5]
rev = []
for i in range(len(l) - 1, -1, -1):
    rev.append(l[i])

print(rev)
```

```python
# another way to reverse
l = [1, 2, 3, 4, 5]
rev = []

for i in l:
    rev.insert(0, i)

print(rev)
```
